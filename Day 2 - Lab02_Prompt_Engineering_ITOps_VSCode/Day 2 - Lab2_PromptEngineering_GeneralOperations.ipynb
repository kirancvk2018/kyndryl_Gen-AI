## 2. What makes a good prompt

A prompt is the instruction you give the model. Vague prompts usually get vague answers.
Most good prompts include some of these four parts:

| Part | The question it answers | Example |
|---|---|---|
| **Role** | Who should the AI act as? | "You are a customer service supervisor at an online store." |
| **Task** | What should it do? | "Decide the next step for this complaint." |
| **Context** | What background does it need? | "The order was due on Monday and has not arrived." |
| **Format** | How should the answer look? | "Reply with three short bullet points." |

We will use one example throughout the lab: a customer complaint about a late delivery.

```python
complaint = """Customer: Maria Lopez
Order #48213 was due on Monday. It is now Thursday and the parcel has not arrived.
The tracking page has shown 'In transit' since Friday. She has emailed twice with no reply.
She is asking for a refund or a replacement."""

print(complaint)
```

## 3. Technique 1: Basic vs. improved prompt

First, a basic prompt with no role, no format and no context.
Run the cell, then read the answer.

```python
basic_prompt = f"""Read this complaint and tell me what to do.

{complaint}"""

ask(basic_prompt)
```

Now the same complaint with an improved prompt that includes a role, a task, context and a format.
Run it and compare the two answers.

```python
improved_prompt = f"""You are a customer service supervisor at an online store.

Task: Decide the next step for the complaint below.
Format: Reply with three short bullet points:
1. What happened
2. What to do today
3. What to say to the customer
Keep the tone calm and apologetic.

Complaint:
{complaint}"""

ask(improved_prompt)
```

**Reflect:** Which answer is easier for a team member to act on? Which one would you send to a customer?

## 4. Technique 2: Giving the model a role

The same complaint can be seen from different roles. A role changes what the model focuses on.
Run the cell below, then change the role in the first line to `"a delivery operations manager at a regional warehouse"` and run it again.

```python
role_prompt = f"""You are a customer service supervisor at an online store.
In 3 bullet points, explain what a customer service team should check before replying to this complaint.

{complaint}"""

ask(role_prompt)
```

## 5. Technique 3: Asking for a specific format

When you need output that someone else can use, such as a task list for a team, ask for a table with fixed columns.
The meeting notes below are from a store manager's weekly meeting.

```python
meeting_notes = """Weekly store meeting, Tuesday.
- Sam will restock the cold-drinks fridge before the weekend rush.
- The back door lock is sticking. Facilities needs to be called.
- Priya is still waiting on the staffing plan for December. Due Friday.
- Everyone should check expiry dates on the bakery shelf daily starting this week.
- Ahmed to order new name badges for the three new hires."""

table_prompt = f"""You are an assistant for a store manager.
Turn the meeting notes below into an action list.
Format: a table with the columns Owner | Action | Due date.
If a field is not in the notes, write "Not stated".

Notes:
{meeting_notes}"""

ask(table_prompt)
```

## 6. Technique 4: Learning from examples (few-shot prompting)

Sometimes the easiest way to get the right answer is to show the model a few examples.
Below, the model learns how to label incoming requests in the examples and then labels a new one.

```python
examples = """Request: The office printer on floor 2 is jammed again.
Category: Facilities | Priority: Medium

Request: A customer says they were charged twice for the same order.
Category: Billing | Priority: High

Request: A new hire starts Monday and needs a desk and an access badge.
Category: HR | Priority: Medium"""

new_request = "The walk-in fridge temperature alarm has been going off since 6 am and the milk is at risk."

few_shot_prompt = f"""Classify each request by Category (Facilities, Billing, HR, Delivery, Other) and Priority (High, Medium, Low).
Follow the pattern in the examples.

Examples:
{examples}

Request: {new_request}
Category:"""

ask(few_shot_prompt)
```

**Reflect:** Change the new request to something that is not in the examples, such as a delivery that arrived damaged. Does the model still follow the pattern?
