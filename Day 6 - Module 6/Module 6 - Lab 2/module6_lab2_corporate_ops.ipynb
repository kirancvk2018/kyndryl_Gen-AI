# Module 6 Lab 2 (Corporate Operations Variant): Sequential LangChain Workflow

This lab builds a four-step sequential chain with LCEL for routine corporate operations requests.

1. **Summary**: what is being asked, category and priority
2. **Cause and blockers**: why the issue happened and what is stopping it
3. **Business impact**: who is affected and what happens if it is late
4. **Action plan**: next steps with owners

Each step is `prompt | llm | StrOutputParser()`. Each step's output is passed to the next step.

## Part 1: Setup

Put your `OPENAI_API_KEY` in a `.env` file in this folder. Install packages with `pip install -r requirements_module6_lab2.txt`.

```python
import os
from dotenv import load_dotenv

load_dotenv()
print('OpenAI API key configured:', 'Yes' if os.getenv('OPENAI_API_KEY') else 'No')
```

## Part 2: Load the Requests

The sample file `data/ops_requests.json` has five routine operations requests.

```python
from ops_workflow import load_requests

requests = load_requests('data/ops_requests.json')
for r in requests:
    print(r['request_id'], '-', r['request_type'], '-', r['department'])
```

## Part 3: Build the Chains

`build_chains()` creates four LCEL chains. Open `ops_workflow.py` to see the prompt templates.

```python
from ops_workflow import build_chains

chains = build_chains(model_name='gpt-4o-mini', temperature=0.3)
print('Chains ready:', list(chains.keys()))
```

## Part 4: Run the Sequential Workflow on One Request

`analyze_request()` runs the four steps in order. Step 2 uses the summary, step 3 uses the summary and cause, and step 4 uses all three.

```python
from ops_workflow import analyze_request, format_report

result = analyze_request(chains, requests[0])
print(format_report(result))
```

## Part 5: Run All Requests and Save

Run the workflow on every request and save the results to JSON.

```python
from ops_workflow import save_results

all_results = [analyze_request(chains, r) for r in requests]
save_results(all_results, 'ops_analysis_results.json')
```
