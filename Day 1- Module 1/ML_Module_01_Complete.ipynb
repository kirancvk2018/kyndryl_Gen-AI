## 1.1 What is Artificial Intelligence?

**Definition:** Artificial Intelligence (AI) is the field of computer science dedicated to creating systems capable of performing tasks that typically require human intelligence.

### AI Characteristics
- **Learning:** Improving performance through experience
- **Reasoning:** Making logical inferences from data
- **Problem-Solving:** Finding optimal solutions
- **Perception:** Recognizing patterns in data
- **Language:** Understanding and generating text

### AI Scope Categories
| Type | Definition | Examples |
|------|-----------|----------|
| **Narrow AI** (Weak AI) | Specialized to specific tasks | Chess engines, spam filters, this course |
| **General AI** (Strong AI) | Human-level intelligence across domains | Theoretical, not yet achieved |
| **Super AI** (ASI) | Exceeds human capability | Hypothetical future |

---

## 1.2 What is Machine Learning?

**Definition:** Machine Learning (ML) is a subset of AI where systems learn patterns from data without explicit programming for every scenario.

### Traditional Programming vs Machine Learning

**Traditional Approach:**
```
Rules (Hardcoded) + Data → Computer → Output
```
Example: IF cpu > 80% THEN alert

**Machine Learning Approach:**
```
Historical Data + Observed Outcomes → ML Algorithm → Learned Rules
```
Example: Learn relationship between 20 metrics and utilization score

### Types of Machine Learning

#### **Supervised Learning** (We use this today)
- Training data includes both inputs (X) and correct outputs (y)
- **Regression:** Predict continuous values (price, temperature, utilization scores)
- **Classification:** Predict categories (fraud/not fraud, healthy/faulty)
- Example: We have 25,000 labeled server records

#### **Unsupervised Learning**
- Only input data (X) - no correct answers provided
- **Clustering:** Group similar items (find server types automatically)
- **Dimensionality Reduction:** Reduce feature count
- Example: Find which servers are similar without being told

#### **Semi-Supervised Learning**
- Mix of labeled and unlabeled data
- Useful when labeling is expensive
- Example: 1,000 labeled records, 24,000 unlabeled

#### **Reinforcement Learning**
- System learns through trial and error with rewards
- Agents take actions to maximize rewards
- Example: Game-playing AI, robotics

## 1.3 Deep Learning vs Traditional ML

**Deep Learning** uses artificial neural networks with multiple layers (hence "deep").

### When to Use Deep Learning
- Very large datasets (millions+ records)
- Unstructured data (images, audio, text)
- Complex, highly non-linear relationships
- Abundant computational resources available

### When to Use Traditional ML (This Course)
- Structured tabular data
- Clear, interpretable relationships
- Limited resources or data
- Fast training and inference required
- **Explainability crucial for business decisions** ← Key for enterprise

### Comparison
| Aspect | Deep Learning | Linear Regression |
|--------|---------------|-------------------|
| **Data Needed** | Millions of records | Thousands of records |
| **Training Time** | Hours/Days/Weeks | Seconds/Minutes |
| **Interpretability** | "Black box" | Fully interpretable |
| **Hardware** | GPUs/TPUs needed | CPU only |
| **Accuracy** | High (90%+) on images | High (85-95%) on tabular |

## 1.4 Generative AI

**Generative AI** systems create new content (text, images, code) based on learned patterns.

### Generative vs Predictive AI

| Aspect | Predictive (Our Focus) | Generative |

## 1.6 Regression Analysis

**Regression** is supervised learning for predicting continuous numerical values.

### Regression Types
| Type | Formula | Use Case | Complexity |
|------|---------|----------|----------|
| **Linear** | y = mx + b | Server utilization (TODAY) | Simple |
| **Polynomial** | y = a + bx + cx² | Non-linear trends | Medium |
| **Ridge/Lasso** | Linear + Penalty | Prevent overfitting | Medium |
| **SVR** | Kernel-based | Complex patterns | High |
| **Random Forest** | Ensemble trees | High accuracy | High |
| **Neural Networks** | Multi-layer | Very complex | Very High |

### Linear Regression Mathematics

**Simple Linear Regression (1 feature):**
$$\hat{y} = \beta_0 + \beta_1 x$$

**Multiple Linear Regression (Many features):**
$$\hat{y} = \beta_0 + \beta_1 x_1 + \beta_2 x_2 + ... + \beta_n x_n$$

Where:
- **ŷ** = Predicted resource utilization score
- **β₀** = Intercept (baseline utilization)
- **β₁, β₂, ..., βₙ** = Coefficients (feature importance weights)
- **x₁, x₂, ..., xₙ** = Input features (CPU, memory, etc.)

## 2.2 Reactive vs Predictive Operations

### Comparison

| Dimension | Reactive | Predictive |
|-----------|----------|------------|
| **Detection** | After failure occurs | Before problem happens |
| **Timeline** | Emergency response (1-24 hrs) | Planned mitigation (days ahead) |
| **Cost** | High (emergency teams, overtime) | Low (automatic, efficient) |
| **User Impact** | Severe (outage/slow service) | None (prevention) |
| **Decision Making** | Manual, ad-hoc | Automated, data-driven |
| **Scalability** | Cannot handle growth | Scales with demand |
| **Team Burden** | Stressful firefighting | Planned, predictable work |

### Workflow Comparison

**REACTIVE WORKFLOW:**
```
Monday (Utilization: 90%)
  ↓
Saturday (Traffic spike)
  ↓
SERVER CRASHES
  ↓
Calls from users
  ↓
Emergency team mobilized (6am Sunday)
  ↓
Emergency provisioning
  ↓
Restore service (4pm Sunday = 10+ hrs downtime)
  ↓
Cost: $50,000+
```

**PREDICTIVE WORKFLOW:**
```
Monday (System predicts: will need +2,000 servers Saturday)
  ↓
Tuesday (Capacity planning approves budget)
  ↓
Wednesday (Auto-scaling provisions +2,000 servers)
  ↓
Thursday-Friday (Test and validate)
  ↓
Saturday (Traffic spike handled smoothly)
  ↓
Zero downtime, no user impact
  ↓
Cost: $5,000 (planned infrastructure)
```

**Savings: $45,000 per incident × 12 incidents/year = $540,000/year**

### **TARGET VARIABLE**

## **resource_utilization_score**
- **Type:** Continuous
- **Range:** 0-100 (percentage)
- **Definition:** Composite metric of overall server resource usage
- **Calculation:** Weighted average combining CPU (35%), Memory (35%), Disk (15%), Network (10%), Health (5%)
- **Business Interpretation:**
  - **0-20:** Severely underutilized (waste)
  - **20-40:** Underutilized
  - **40-60:** Optimal range (target)
  - **60-80:** High utilization (monitor)
  - **80-100:** Critical (risk)

### **Our Prediction Task**
Use 26 independent variables (features) to predict resource_utilization_score

## 3.3 Variable Classifications

### **Continuous (Numerical) Variables**
Can take any value within a range. Used directly in Linear Regression.

Percentage metrics: cpu_usage, memory_usage, disk_usage, bandwidth_utilization, packet_loss_pct

Count metrics: disk_io, running_processes, failed_processes, active_users, security_alerts

Time/Physical: network_latency_ms, temperature_c, fan_speed_rpm, power_consumption_w, uptime_days, patch_age_days, maintenance_due_days

Target: resource_utilization_score

### **Categorical Variables**
Represent categories. Must be encoded for algorithms.

| Variable | Categories | Encoding Strategy |
|----------|------------|------------------|
| **datacenter** | dc_us, dc_eu, dc_asia | One-Hot Encoding |
| **region** | us_east, eu_west, asia_se | One-Hot Encoding |
| **operating_system** | Windows, Linux, Unix | One-Hot Encoding |
| **server_type** | database, web, application, cache, batch | One-Hot Encoding |
| **workload_type** | compute, io, memory, balanced | One-Hot Encoding |
| **backup_status** | success, in_progress, failed | One-Hot Encoding |
| **virtualization_host** | yes, no | Label Encoding (0, 1) |
| **service_priority** | critical, high, medium, low | Label Encoding (3, 2, 1, 0) |
